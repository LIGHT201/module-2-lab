```python
# Brijesh Patel
# student_qualification.py

# This program asks for the first name of the user or to input 'zzz'
# which is turned to lowercase then asks for the user's last name.
# next it asks for the users g


while (student_first_name := input("Input First Name or type 'ZZZ' to exit: ")).upper() != 'ZZZ':
    student_last_name = input("Enter Last name: ")
    gpa = float(input("Enter student's GPA: "))
    
    if gpa >= 3.5:
        print(f"{student_first_name} {student_last_name} has made the Dean's List!")
    elif gpa >= 3.25:
        print(f"{student_first_name} {student_last_name} has made the Honor Roll!")
    else:
        print(f"{student_first_name} {student_last_name} did not make it to Honor Roll nor the Dean's List.")
```
